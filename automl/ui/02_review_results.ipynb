Open the successful run recorded in [01 — Create an experiment](01_create_experiment.ipynb). Review the run's actual input data and metric rather than assuming it used the latest form settings. No code-route manifest or `run.json` is required for this UI route.

## 1. Read the leaderboard

1. Open the run's AutoML results page and check its run ID and selected metric.
2. Inspect the leaderboard and compare candidate scores using the configured metric, initially ROC-AUC.
3. Record the selected predictor's name, score, and any available F1, precision, recall, and accuracy values.
4. Open the existing pipeline run to inspect per-model metrics/artifacts if additional details are needed. Compare the same metric across candidates.

Accuracy alone can hide missed fraud. Native AutoML UI runs use the pipeline's internal evaluation split, not the repository's external `data/test.csv`. These scores participate in model selection and are not an untouched final generalization estimate. A preset comparison is meaningful only when training input, metric, and evaluation protocol match.

## 2. Inspect the selected predictor and evaluation results

1. Open the selected candidate's details where available. Record its name, score, and artifact location.
2. Record that this experiment was given only the training dataset. The managed pipeline produced the evaluation results automatically; no separate evaluation/test CSV was supplied.
3. Inspect any available evaluation artifacts and record only the statistics actually exposed. Do not copy the repository's `data/test.csv` row counts or fraud rate into this run's results.
4. For the selected predictor, open its model details and choose **View → Save as notebook** where available. Download the run-level experiment notebook using the run's notebook/download controls, or find both official notebooks in the artifact bucket through the storage console. Open artifacts from your trusted run and follow their dependency instructions if you execute them.

Generated notebooks are an optional bridge to local experimentation and `predict_proba`. Executing them uses code; the dashboard walkthrough can continue directly to deployment.

## 3. Locate the complete predictor directory

The model-details **View → Save as notebook** action is shown below. To record the predictor directory as well, use the model's artifact location shown in the results, the run's output artifacts, or the object-storage console. The managed pipeline stores per-model artifacts beneath the run prefix; take the actual location rather than guessing a task ID.

For deployment you need the **whole directory ending in `predictor/`**, not a single `predictor.pkl` file. It can contain dependent models and supporting files. The directory can belong to a stacked predictor whose name alone does not describe every dependency.

| Item | Record from the run |
| --- | --- |
| Selected model | `<SELECTED_MODEL>` |
| Metric / score | `<EVAL_METRIC>` / `<OBSERVED_SCORE>` |
| Evaluation source / rows / verification performed | `Automatic evaluation from training input` / `<OBSERVED_ROWS_OR_NOT_EXPOSED>` / `<CHECK_PERFORMED>` |
| Full predictor URI | `s3://<YOUR_ARTIFACT_BUCKET>/<ACTUAL_PREDICTOR_PREFIX>/predictor/` |
| Deployment path within bucket | `<ACTUAL_PREDICTOR_PREFIX>/predictor` |
| Artifact-bucket Connection | `<YOUR_ARTIFACT_CONNECTION>` |
| Official experiment / predictor notebook location | `<ACTUAL_NOTEBOOK_LOCATIONS>` |

**Ready to continue:** the model files exist in the recorded directory, and the deployment Connection points at that artifact bucket. The pipeline output bucket can differ from the training-input bucket.
