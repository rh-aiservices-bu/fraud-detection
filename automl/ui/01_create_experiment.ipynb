Use the project, Connection, bucket, and training object recorded in [00 — Data and Connections](00_setup_and_data.ipynb). Creating an experiment submits real cluster work; inspect the review screen before clicking the final create/run action.

**Dataset input in this release: `train.csv` only.** Separate evaluation/test data cannot be uploaded or selected. The managed pipeline handles evaluation automatically. Support for supplying evaluation data is planned for later releases.

## 1. Start a tabular AutoML experiment

1. Open the dashboard's **AutoML** area under **Develop and train** and select your project.
2. Choose the action to create an experiment and select tabular data / classification where offered.
3. Give the experiment a descriptive name, for example `fraud-detection-automl`.

The exact button and field labels depend on the installed version. If AutoML is missing, ask the administrator to check feature enablement and managed pipeline availability. Use only the training dataset selector. There is no evaluation/test dataset control in this release.

## 2. Select the training source

1. Select the input Connection from notebook 00.
2. Enter/select the training CSV object, using the recorded bucket and key. If the form asks for a full URI, use `s3://<YOUR_INPUT_BUCKET>/<TRAIN_OBJECT_KEY>`; if it asks for a path within the bucket, enter the key only.
3. Select **fraud** as the target/label column, **binary classification** as the task, and **1** as the positive class where exposed.
4. Keep all seven feature columns. Do not include `fraud` as an input feature or reuse the original five-feature ONNX input format.

If a schema preview is offered, check the target and all seven features before continuing.

## 3. Configure search settings

| Setting | Workshop choice | Purpose |
| --- | --- | --- |
| Managed preset | `speed` | Start with the GA speed preset; its selection budget is 45 minutes, plus refit, evaluation, and scheduling |
| Evaluation/ranking metric | `roc_auc` where selectable | Prefer ROC-AUC or F1 on this imbalanced dataset; record the actual metric used by your UI run |
| Positive class | `1` where exposed | Fraud is the positive outcome |
| Top N | `1` where exposed | Record the installed form's choice/default; the managed pipeline supports 1–10 |

Review the controls available in your installed UI and record their actual values/defaults. A `speed` run and a `balanced` comparison should have separate names and run IDs, with the same evaluation protocol for a meaningful comparison.

## 4. Review, submit, and monitor

1. Review the project, Connection, training object, label, task type, preset, and metric.
2. Confirm that this native UI run uses internal evaluation, then submit **once** and record the run ID, pipeline version, and start time.
3. Monitor status in AutoML and, where available, open the run in **Pipelines → Runs** for task details and logs.
4. Wait for success before reviewing scores. Closing this notebook/browser does not cancel training; return to the existing run. Use the run's stop/cancel action to cancel unwanted work.
5. For a failure, inspect task logs and pod termination reasons. Record `OOMKilled` only when observed; a failed status alone does not establish OOM.

Record the links to both views. The dashboard routes used by the code sample are `/develop-train/automl/results/<YOUR_PROJECT>/<RUN_ID>` and `/develop-train/pipelines/runs/<YOUR_PROJECT>/runs/<RUN_ID>` under your dashboard base URL.

## 5. Record the actual run

| Evidence | Your value |
| --- | --- |
| Run ID / pipeline version | `<RUN_ID>` / `<PIPELINE_VERSION_ID>` |
| AutoML results URL | `<AUTOML_RESULTS_URL>` |
| Pipeline run details URL | `<RUN_DETAILS_URL>` |
| Training dataset URI | `<TRAIN_URI>` |
| Dataset input | `Training CSV only` |
| Evaluation handling | `Automatic, from the training input; no evaluation dataset supplied` |
| Actual preset / metric / top N, where exposed | `<PRESET>` / `<METRIC>` / `<TOP_N>` |
| Status / duration | `<STATE>` / `<DURATION>` |
| Failure/OOM evidence, if any | `<OBSERVED_EVIDENCE>` |

Open this run's details to inspect its recorded configuration without submitting another run. **Ready to continue:** the recorded run succeeded. Changing form values later does not change that run; record a new run ID for any comparison experiment.
